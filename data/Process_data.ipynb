<h2>Prepare data</h2>

```python
df_en_et = pd.read_csv("./en-et-2024.csv", header=None, names=['English', 'Estonian'])
df_en_sv = pd.read_csv("./en-sv-2024.csv", header=None, names=['English', 'Swedish'])
```

```python
df_en_et = df_en_et.dropna()
df_en_sv = df_en_sv.dropna()
```

```python
df_en_et = df_en_et.drop_duplicates()
df_en_sv = df_en_sv.drop_duplicates()
```

```python
df_en_et = df_en_et.drop_duplicates(subset='English', keep='first')
df_en_sv = df_en_sv.drop_duplicates(subset='English', keep='first')
```

```python
df = pd.merge(df_en_et, df_en_sv, on='English', how='inner')
```

```python
df.shape
```

```python
df.head(20)
```

```python
df_sample = df.sample(n=10000, random_state=42)
```

```python
df_sample = df_sample.map(lambda x: x[1:].strip() if isinstance(x, str) and x.startswith('=') else x)
```

```python
df_sample.shape
```

```python
df_sample.reset_index(inplace=True, drop=True)
```
